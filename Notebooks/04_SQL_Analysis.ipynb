## Create an In-Memory SQL Database

The cleaned dataset is loaded into an SQLite database to perform SQL analysis.

```python
conn = sqlite3.connect(":memory:")

df.to_sql(
    "employees",
    conn,
    if_exists="replace",
    index=False
)
```

## What Is the Employee Attrition Rate?

This analysis calculates the number of employees who stayed and those who left the company.

```python
query = """
SELECT
    Attrition,
    COUNT(*) AS employee_count
FROM employees
GROUP BY Attrition;
"""

pd.read_sql(query, conn)
```

## Business Insight

Employee attrition directly affects productivity, recruitment costs, and employee retention.

Understanding attrition rates helps organizations develop effective retention strategies.

## Which Departments Have the Highest Attrition?

This analysis identifies departments with the highest number of employees leaving the organization.

```python
query = """
SELECT
    Department,
    COUNT(*) AS attrition_count
FROM employees
WHERE Attrition='Yes'
GROUP BY Department
ORDER BY attrition_count DESC;
"""

pd.read_sql(query, conn)
```

## Which Job Roles Have the Highest Attrition?

This analysis identifies job roles with the highest employee turnover.

```python
query = """
SELECT
    JobRole,
    COUNT(*) AS attrition_count
FROM employees
WHERE Attrition='Yes'
GROUP BY JobRole
ORDER BY attrition_count DESC;
"""

pd.read_sql(query, conn)
```

## Does Overtime Increase Employee Attrition?

This analysis compares attrition among employees who work overtime and those who do not.

```python
query = """
SELECT
    OverTime,
    COUNT(*) AS attrition_count
FROM employees
WHERE Attrition='Yes'
GROUP BY OverTime;
"""

pd.read_sql(query, conn)
```

## Which Age Groups Experience the Highest Attrition?

This analysis identifies the age groups with the highest employee turnover.

```python
query = """
SELECT
    Age,
    COUNT(*) AS attrition_count
FROM employees
WHERE Attrition = 'Yes'
GROUP BY Age
ORDER BY attrition_count DESC;
"""

pd.read_sql(query, conn)
```

## Does Salary Affect Employee Attrition?

This analysis compares the average monthly income of employees who stayed and those who left the company.

```python
query = """
SELECT
    Attrition,
    ROUND(AVG(MonthlyIncome), 2) AS average_income
FROM employees
GROUP BY Attrition;
"""

pd.read_sql(query, conn)
```

## Which Education Fields Have the Highest Attrition?

This analysis identifies education fields associated with the highest employee turnover.

```python
query = """
SELECT
    EducationField,
    COUNT(*) AS attrition_count
FROM employees
WHERE Attrition = 'Yes'
GROUP BY EducationField
ORDER BY attrition_count DESC;
"""

pd.read_sql(query, conn)
```

## Which Marital Status Groups Have the Highest Attrition?

This analysis examines the relationship between marital status and employee attrition.

```python
query = """
SELECT
    MaritalStatus,
    COUNT(*) AS attrition_count
FROM employees
WHERE Attrition = 'Yes'
GROUP BY MaritalStatus
ORDER BY attrition_count DESC;
"""

pd.read_sql(query, conn)
```

## Which Business Travel Category Has the Highest Attrition?

This analysis examines whether business travel influences employee turnover.

```python
query = """
SELECT
    BusinessTravel,
    COUNT(*) AS attrition_count
FROM employees
WHERE Attrition = 'Yes'
GROUP BY BusinessTravel
ORDER BY attrition_count DESC;
"""

pd.read_sql(query, conn)
```

## Which Employees Have the Longest Tenure?

This analysis identifies employees with the highest number of years at the company.

```python
query = """
SELECT
    JobRole,
    MAX(YearsAtCompany) AS longest_tenure
FROM employees
GROUP BY JobRole
ORDER BY longest_tenure DESC;
"""

pd.read_sql(query, conn)
```

## Business Insights

Key findings from the SQL analysis include:

- Employee attrition varies across departments.
- Certain job roles experience higher turnover.
- Overtime appears to influence employee attrition.
- Employees with lower salaries may be more likely to leave.
- Business travel may contribute to employee turnover.
- Education level and marital status may influence retention.
- Employee tenure differs significantly across job roles.

## Objective

The objective of this notebook is to clean and prepare the HR Employee Attrition dataset for analysis.

This notebook focuses on:

- Identifying duplicate records
- Handling missing values
- Verifying data types
- Removing unnecessary columns
- Preparing the dataset for exploratory data analysis

## Import Required Libraries

The Pandas library is used to load and analyze the HR Employee Attrition dataset.

```python
import pandas as pd
```

```python
df = pd.read_csv("../Data/01_raw/hr_employee_attrition.csv")
```

## Dataset Copy

A copy of the original dataset is created to preserve the raw data.

```python
df_clean = df.copy()
```

## Check for Duplicate Records

Duplicate records can affect the accuracy of the analysis.

```python
df_clean.duplicated().sum()
```

## Check for Missing Values

Missing values can lead to inaccurate analysis and should be identified before proceeding.

```python
df_clean.isnull().sum()
```

## Verify Data Types

The dataset's data types are examined to ensure they are appropriate for analysis.

```python
df_clean.dtypes
```

## Remove Unnecessary Columns

Some columns do not contribute meaningful information and may be removed.

```python
df_clean.columns.tolist()
```

## Verify Column Values

The following columns are examined to confirm whether they contain constant values or employee identifiers before removing them.

```python
print("StandardHours:", df_clean["StandardHours"].unique())
```

```python
print("EmployeeCount:", df_clean["EmployeeCount"].unique())
```

```python
print("Over18:", df_clean["Over18"].unique())
```

```python
print(df_clean["EmployeeNumber"].head())
```

## Remove Unnecessary Columns

The following columns are removed because they do not provide meaningful information for the analysis:

- EmployeeCount
- EmployeeNumber
- Over18
- StandardHours

```python
columns_to_remove = [
    "EmployeeCount",
    "EmployeeNumber",
    "Over18",
    "StandardHours"
]

df_clean = df_clean.drop(columns=columns_to_remove)
```

```python
df_clean.columns.tolist()
```

## Dataset Shape After Cleaning

The dataset size is checked again after removing unnecessary columns.

```python
print(f"Rows: {df_clean.shape[0]}")
print(f"Columns: {df_clean.shape[1]}")
```

## Save the Cleaned Dataset

The cleaned dataset is saved for use in the exploratory data analysis, SQL analysis, and Power BI dashboard notebooks.

```python
df_clean.to_csv(
    "../Data/02_cleaned/hr_employee_attrition_cleaned.csv",
    index=False
)
```

## Data Cleaning Summary

The following cleaning tasks were completed:

- Duplicate records were checked.
- Missing values were examined.
- Data types were verified.
- Unnecessary columns were removed.
- The cleaned dataset was saved for future analysis.

## Business Insight

The dataset required minimal cleaning because no missing values or duplicate records were detected.

Several columns containing constant values or employee identifiers were removed because they do not contribute to employee attrition analysis.

## Key Business Findings

### Employee Attrition

- The organization employs 1,470 employees.

- A total of 237 employees left the company.

- The overall employee attrition rate is 16.12%.

### Department Analysis

- Research & Development is the largest department (961 employees) and has the highest number of departures in absolute terms (133), giving it an attrition rate of 13.8%.

- Sales is the second-largest department (446 employees), with 92 departures — but a higher attrition rate of 20.6%, meaning Sales loses a larger share of its workforce despite being smaller than R&D.

- Human Resources has the smallest workforce (63 employees) and 12 departures, an attrition rate of roughly 19%.

- Key insight: attrition rate, not just raw headcount lost, reveals Sales as the department under the most retention pressure relative to its size.

### Job Role Analysis

- Laboratory Technicians have the highest attrition count of any job role (62), followed by Sales Executives (57) and Research Scientists (47).

- These three roles account for a disproportionate share of total attrition and should be prioritized for retention initiatives.

### Overtime Analysis

- 71.7% of employees in this dataset work overtime.

- Employees who work overtime are more likely to leave the company, making overtime one of the strongest and most actionable factors influencing attrition.

### Compensation Analysis

- Average monthly income across the organization is 6,503.

- Monthly income varies significantly across employees and may influence satisfaction and retention, particularly in high-attrition roles like Sales Executive and Laboratory Technician.

### Business Travel Analysis

- Employees who travel rarely make up the largest share of attrition cases (156), likely reflecting that most employees fall into this travel category overall.

- Frequent travelers show a meaningfully elevated attrition rate relative to their smaller group size (69 departures), suggesting travel frequency remains a contributing stress factor worth monitoring.

### Work-Life Balance Analysis

- Most employees report a work-life balance rating of 3 (893 employees), the largest group in the dataset.

- Employees with lower work-life balance ratings (1 and 2) show notably smaller representation, warranting further investigation into whether this correlates with attrition risk.

## Recommendations

### Reduce Overtime

- Monitor employee workloads.

- Introduce flexible work schedules.

- Improve workforce planning to reduce excessive overtime.

### Improve Employee Retention

- Develop retention programs for high-risk job roles.

- Create employee engagement initiatives.

- Conduct regular employee satisfaction surveys.

### Review Compensation Policies

- Evaluate salary structures.

- Compare employee compensation across departments and job roles.

- Introduce performance-based incentives.

- Benchmark salaries against industry standards.

### Improve Work-Life Balance

- Expand flexible working arrangements.

- Encourage employees to use vacation time.

- Support employee wellness programs.

### Review Business Travel Policies

- Reduce unnecessary travel.

- Offer hybrid or virtual alternatives when possible.

- Provide additional support for employees who travel frequently.

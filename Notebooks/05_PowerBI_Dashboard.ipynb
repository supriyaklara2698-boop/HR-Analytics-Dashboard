## Business Insights

The Power BI dashboard revealed several important findings:

- The company has 1,470 employees, and 237 employees have left the organization.

- The employee attrition rate is 16.12%, indicating that employee turnover is a significant business challenge.

- Research and Development is the largest department, employing 961 employees.

- Sales Executives, Research Scientists, and Laboratory Technicians experience the highest levels of employee attrition.

- Employees who frequently travel for business appear to leave the company more often than employees who do not travel.

- Employees who work overtime represent a large portion of the employees who leave the organization.

- Most employees report a work-life balance rating of 3, suggesting an acceptable but improvable work environment.

- Education background, compensation, overtime, and department assignment appear to influence employee retention.

- Attrition appears to be concentrated in specific departments and job roles rather than being evenly distributed across the organization.

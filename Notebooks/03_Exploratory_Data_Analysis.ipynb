## Objective

The objective of this notebook is to explore the HR Employee Attrition dataset and identify patterns that may influence employee turnover.

This notebook focuses on:

- Understanding employee attrition patterns
- Exploring employee demographics
- Analyzing compensation
- Examining employee satisfaction
- Identifying relationships between important variables

## Import Required Libraries

The following libraries are used for data analysis and visualization.

- Pandas
- Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../Data/02_cleaned/hr_employee_attrition_cleaned.csv")
```

## Dataset Preview

The cleaned dataset is loaded to begin exploratory data analysis.

```python
df.head()
```

## Employee Attrition Distribution

This analysis examines the number of employees who stayed and those who left the company.

```python
df["Attrition"].value_counts()
```

```python
df["Attrition"].value_counts().plot(kind="bar")

plt.title("Employee Attrition Distribution")

plt.xlabel("Attrition")

plt.ylabel("Number of Employees")

plt.show()
```

## Employees by Department

This analysis examines the distribution of employees across departments.

```python
df["Department"].value_counts()
```

```python
df["Department"].value_counts().plot(kind="bar")

plt.title("Employees by Department")

plt.xlabel("Department")

plt.ylabel("Number of Employees")

plt.show()
```

## Department Insight

Research and Development contains the largest number of employees.

Understanding department distribution helps identify areas where employee attrition may have the greatest impact.

## Employees by Job Role

This analysis identifies the most common job roles in the organization.

```python
df["JobRole"].value_counts()
```

```python
df["JobRole"].value_counts().plot(kind="bar")

plt.title("Employees by Job Role")

plt.xlabel("Job Role")

plt.ylabel("Number of Employees")

plt.xticks(rotation=90)

plt.show()
```

## Job Role Insight

Laboratory Technicians and Sales Executives represent a significant portion of the workforce.

Job role distribution may influence employee turnover patterns.

## Monthly Income Distribution

This analysis examines employee salary distribution.

```python
df["MonthlyIncome"].describe()
```

```python
plt.hist(df["MonthlyIncome"], bins=20)

plt.title("Monthly Income Distribution")

plt.xlabel("Monthly Income")

plt.ylabel("Frequency")

plt.show()
```

## Monthly Income Insight

Employee salaries vary considerably across the organization.

The salary distribution is not uniform, which suggests that compensation may differ significantly across job roles and experience levels.

Monthly income may be an important factor influencing employee attrition.

## Overtime Distribution

This analysis examines how many employees work overtime.

```python
df["OverTime"].value_counts()
```

```python
df["OverTime"].value_counts().plot(kind="bar")

plt.title("Overtime Distribution")

plt.xlabel("Overtime")

plt.ylabel("Number of Employees")

plt.show()
```

## Overtime Insight

Employees who work overtime may experience higher levels of stress and work-related pressure.

Overtime is a potential factor that may influence employee attrition.

## Work-Life Balance Distribution

This analysis examines employee work-life balance ratings.

```python
df["WorkLifeBalance"].value_counts().sort_index()
```

```python
df["WorkLifeBalance"].value_counts().sort_index().plot(kind="bar")

plt.title("Work-Life Balance Distribution")

plt.xlabel("Work-Life Balance")

plt.ylabel("Number of Employees")

plt.show()
```

## Business Insights

Key findings from the exploratory analysis include:

- Employee attrition affects a significant portion of the workforce.
- Research and Development is the largest department.
- Some job roles have more employees than others.
- Monthly income varies considerably across employees.
- Overtime may influence employee attrition.
- Work-life balance may affect employee retention.
- Compensation may influence employee retention and job satisfaction.
